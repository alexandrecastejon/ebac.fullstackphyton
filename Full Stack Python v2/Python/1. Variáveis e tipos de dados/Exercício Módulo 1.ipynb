## 3\. Strings

Aplique três métodos distintos na *string* abaixo, você pode conferir alguns métodos neste [link](https://www.w3schools.com/python/python_ref_string.asp):

```
cancao = 'Roda mundo, roda gigante, roda moinho, roda pião.'

# Método 1: Converter para maiúsculas (upper())
cancao_maiuscula = cancao.upper()
print(f"Original: {cancao}")
print(f"Maiúscula: {cancao_maiuscula}\n")

# Método 2: Substituir uma substring (replace())
cancao_substituida = cancao.replace('roda', 'gira')
print(f"Substituindo 'roda' por 'gira': {cancao_substituida}\n")

# Método 3: Dividir a string em uma lista (split())
cancao_dividida = cancao.split(', ')
print(f"Dividida por ', ': {cancao_dividida}")
```

Extraia da string abaixo o valor da taxa **selic** na variável `selic` e o valor do **ano** na variavel `ano`. Imprima os valores na tela.

```
noticia = 'Selic vai a 2,75% e supera expectativas; é a primeira alta em 6 anos.'

# Extrair o valor da Selic
start_selic = noticia.find('Selic vai a ') + len('Selic vai a ')
end_selic = noticia.find('%', start_selic)
selic = noticia[start_selic:end_selic+1]

# Extrair o valor do ano
start_ano = noticia.find('em ') + len('em ')
end_ano = noticia.find(' anos', start_ano)
ano = noticia[start_ano:end_ano]

print(f"Taxa Selic: {selic}")
print(f"Ano: {ano}")
```
