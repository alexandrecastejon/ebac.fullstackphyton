## 1\. Funções para arquivo csv

Complete a função abaixo para extrair uma coluna do arquivo csv em uma lista. Os elementos devem ter o tipo de dado correto.

```
def extrai_coluna_csv(nome_arquivo: str, indice_coluna: int, tipo_dado: str):

  coluna = []

  with open(nome_arquivo, 'r') as arquivo:
    # Skip header
    next(arquivo) # Pula a linha do cabeçalho
    for linha in arquivo:
      elementos = linha.strip().split(',')
      dado = elementos[indice_coluna]

      if tipo_dado == 'str':
        coluna.append(str(dado))
      elif tipo_dado == 'int':
        coluna.append(int(dado))
      elif tipo_dado == 'float':
        coluna.append(float(dado))
      else:
        # Caso o tipo de dado não seja reconhecido, adiciona como string por padrão
        coluna.append(str(dado))

  return coluna
```

Você pode testar a função com os códigos abaixo.

```
# extrair a coluna valor_venda
valor_venda = extrai_coluna_csv(nome_arquivo='./carros.csv', indice_coluna=1, tipo_dado='str')
print(valor_venda) # deve retornar ['vhigh', 'med', 'low', ...]
```

```
# extrair a coluna pessoas
pessoas = extrai_coluna_csv(nome_arquivo='./carros.csv', indice_coluna=4, tipo_dado='int')
print(pessoas) # deve retornar [2, 2, 2, ...]
```

## 2\. Funções para arquivo txt

Complete a função abaixo para extrair uma as palavras de uma linha do arquivo txt em uma lista.

```
def extrai_linha_txt(nome_arquivo: str, numero_linha: int):

  palavras_linha = []

  with open(nome_arquivo, 'r') as arquivo:
    linhas = arquivo.readlines()
    if 0 < numero_linha <= len(linhas):
      linha_selecionada = linhas[numero_linha - 1].strip() # -1 because lists are 0-indexed
      palavras_linha = linha_selecionada.split(' ')

  return palavras_linha
```

Você pode testar a função com os códigos abaixo.

```
linha10 = extrai_linha_txt(nome_arquivo='./musica.txt', numero_linha=10)
print(linha10) # deve retornar ['Mas', 'eis', 'que', 'chega', 'a', 'roda', 'viva']
```
